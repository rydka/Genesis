## 2. Import Libraries and Initialize

```python
import sys
sys.path.append('.')

from Genesis import Genesis3D, ViTDepthEstimator
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import torch

print(f"CUDA available: {torch.cuda.is_available()}")
print(f"Device: {torch.device('cuda' if torch.cuda.is_available() else 'cpu')}")
```

## 3. Initialize Genesis 3D System

```python
# Initialize the Genesis 3D system
genesis = Genesis3D()
print("Genesis 3D system initialized successfully!")
```

## 4. Test with Sample Image

```python
# Replace with your image path
image_path = "path/to/your/test_image.jpg"

# Check if image exists
import os
if os.path.exists(image_path):
    print(f"Image found: {image_path}")
else:
    print(f"Image not found: {image_path}")
    print("Please update the image_path variable with a valid image file")
```

## 5. Generate Depth Map and Visualize

```python
# Load and display original image
if os.path.exists(image_path):
    image = Image.open(image_path).convert("RGB")
    plt.figure(figsize=(8, 6))
    plt.imshow(image)
    plt.title('Original Image')
    plt.axis('off')
    plt.show()
    
    print(f"Image size: {image.size}")
```

```python
# Generate depth map
if os.path.exists(image_path):
    print("Generating depth map...")
    depth_map = genesis.depth_estimator.estimate_depth(image)
    print(f"Depth map shape: {depth_map.shape}")
    print(f"Depth range: {depth_map.min():.3f} to {depth_map.max():.3f}")
```

```python
# Display image and depth map side by side
if os.path.exists(image_path):
    genesis.display_image_and_depth(image, depth_map)
```

## 6. Generate 3D Mesh

```python
# Generate full 3D mesh
if os.path.exists(image_path):
    print("Generating 3D mesh...")
    mesh = genesis.image_to_3d(image_path, visualize=False)
    
    print(f"Mesh vertices: {len(mesh.vertices)}")
    print(f"Mesh triangles: {len(mesh.triangles)}")
    print("3D mesh generated successfully!")
```

## 7. Visualize Point Cloud

```python
# Visualize the point cloud
if os.path.exists(image_path) and hasattr(genesis, 'point_cloud'):
    print("Visualizing point cloud...")
    print(f"Point cloud has {len(genesis.point_cloud.points)} points")
    
    # This will open a separate window
    import open3d as o3d
    o3d.visualization.draw_geometries([genesis.point_cloud], 
                                    window_name="Genesis 3D Point Cloud")
```

## 8. Visualize 3D Mesh

```python
# Visualize the 3D mesh
if os.path.exists(image_path) and genesis.mesh is not None:
    print("Visualizing 3D mesh...")
    
    # This will open a separate window
    o3d.visualization.draw_geometries([genesis.mesh], 
                                    window_name="Genesis 3D Mesh")
```

## 9. Save Results

```python
# Save depth visualization
if os.path.exists(image_path):
    output_viz = "depth_visualization.png"
    genesis.save_depth_visualization(image_path, output_viz)
    print(f"Depth visualization saved to: {output_viz}")
    
    # Save 3D mesh
    if genesis.mesh is not None:
        mesh_output = "generated_mesh.ply"
        o3d.io.write_triangle_mesh(mesh_output, genesis.mesh)
        print(f"3D mesh saved to: {mesh_output}")
        
        # Save point cloud
        if hasattr(genesis, 'point_cloud'):
            pcd_output = "point_cloud.ply"
            o3d.io.write_point_cloud(pcd_output, genesis.point_cloud)
            print(f"Point cloud saved to: {pcd_output}")
```

## 10. Test Different Images

```python
# Test with multiple images
test_images = [
    "path/to/image1.jpg",
    "path/to/image2.jpg",
    "path/to/image3.jpg"
]

for i, img_path in enumerate(test_images):
    if os.path.exists(img_path):
        print(f"\nProcessing image {i+1}: {img_path}")
        mesh = genesis.image_to_3d(img_path, visualize=True)
        
        # Save results
        genesis.save_depth_visualization(img_path, f"depth_viz_{i+1}.png")
        if mesh is not None:
            o3d.io.write_triangle_mesh(f"mesh_{i+1}.ply", mesh)
    else:
        print(f"Image not found: {img_path}")
```
